```python
def create_profile(df: pd.DataFrame) -> pd.DataFrame: 
    # def funcion(nombre_parametro: TipoEsperado) -> TipoQueDevuelve
    n_rows = len(df)
    cardinality = df.nunique(dropna=False)
    nulls = df.isna().sum()
        

    profile = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "cardinality": cardinality,
        "unique_pct": (cardinality/n_rows*100).round(2),
        "nulls": df.isna().sum(),
        "null_pct": (df.isna().mean()*100).round(2),
        "duplicates": df.duplicated().sum(),
    })

    return profile
```

## Data coverage assesment

The coverage tables show whether a valid observation exists for each expected country-year combination and, for activity indicators, each country-year-diagnosis combination. A value of `1` indicates an available observation; `0` indicates that Eurostat does not provide an observation for that combiantion.


```python
beds.sample(5)
```

```python
beds_coverage = (
    beds
    .pivot_table(
        index="geo",
        columns="TIME_PERIOD",
        values="OBS_VALUE",
        aggfunc="count"
    )
    .fillna(0)  
    .astype(int)  # method chaining
)

beds_coverage
```

```python
discharges_coverage = (
    discharges
    .pivot_table(
        index=["geo", "icd10"],
        columns="TIME_PERIOD",
        values="OBS_VALUE",
        aggfunc="count"
    )
    .fillna(0)
    .astype(int)
)

discharges_coverage
```

```python
length_stay_coverage = (
    length_stay
    .pivot_table(
        index=["geo", "icd10"],
        columns="TIME_PERIOD",
        values="OBS_VALUE",
        aggfunc="count"
    )
    .fillna(0)
    .astype(int)
)

length_stay_coverage
```

- Hospital beds have complete coverage for Germany, Spain, France, Italy and Portugal from 2014 to 2019.
- Hospital discharges and average length of stay: cover a wider historical period (2014-21) bur their availability is not fully balanced across all countries and diagnoses. Spain lacks the all-diseases observation in 2014 and Italy from 2019 onwards. Portugal has no observations for any diagnosis in 2016-17.
- The **project analysis period** is therefore set to **2014-19**, which aligns with the complete historical hospital-beds series. Missing observations will be retained as absent data and documented; they will not be imputed.


## Profiling conclusions  

### 1. What does each row represent?  

- `beds`: one observation per reporting country and year, representing total hospital beds in hospitals per 100,000 inhabitants.
- `discharges`: one observation per reporting country, year and ICD-10 diagnosis group, representing in-patient hospital discharges per 100,000 inhabitants.
- `lenght_stay`: one observation per reporting country, year and ICD-10 diagnosis group, representing the average lenght of stay for in-patients in days. 

### 2. Which columns uniquely identify an observation?

The raw datasets contain several metadata columns that are constant after filtering. The analytical keys are: 

- `beds`: `geo`+ `TIME_PERIOD`
- `discharges`: `geo`+ `TIME_PERIOD` + `icd-10`
- `lenght_stay`: `geo`+ `TIME_PERIOD` + `icd-10`

These combinations must be unique after cleaning and will define the granularity of the fact tables. 

### 3. Are years stored in rows or columns?

Years are stored in rows through the `TIME_PERIOD`column. The coverage tables pivot years into columns only to make missing country-year combinations easier to inspect. 

### 4. How are missing values represented?

There are no null values in `OBS_VALUE`in the downloaded records. However, some expected country-year-diagnosis combinations are absent entirely from the source data. These structural absences are identified through the coverage tables and must not be treated as zero values. 

### 5. Is there a separate observation-status flag?

Yes. Eurostat provides `OBS_FLAG`and a corresponding descriptive label column. Only in `beds`, one observation has a status flag. 

The raw `OBS_FLAG` column will be retained initially as `status_flag` where it contains information. Empty flag columns will be removed from the final processed datasets. 

### 6. Are the rates already calculated per 100,000 inhabitants?

- Hospital beds and hospital discharges are already reported per 100,000
  inhabitants.
- Average length of stay is not a population rate. It is an average duration in
  days per hospital discharge.

Therefore, beds and discharges must not be summed across countries or years,
and average length of stay must not be summed across diagnoses, countries or
years.

### 7. Are there multiple units within a dataset?

No. Each downloaded dataset has one unit after filtering.

- Beds: per 100,000 inhabitants.
- Discharges: per 100,000 inhabitants.
- Average length of stay: days.

### 8. What diagnosis codes are included?

The selected diagnosis groups are:

- `A-T_Z`: All diseases
- `C00-D48`: Neoplasms
- `F`: Mental and behavioural disorders
- `I`: Diseases of the circulatory system
- `J`: Diseases of the respiratory system

### 9. Do totals and subcategories create a risk of double counting?

Yes. `A-T_Z` represents all diseases, while `C00-D48`, `F`, `I` and `J`
represent diagnostic subgroups included within the total.

The total must never be reconstructed by summing the selected diagnoses, and
total observations must not be added to subgroup observations.

### 11. What is the final time scope?

The latest year shared by the three source datasets is 2019 because the
hospital-beds historical dataset ends in that year.

The project will analyse 2014–2019. Comparisons involving Portugal in 2016 and
2017, and all-diseases discharge comparisons where coverage is incomplete,
will be excluded or explicitly labelled as incomplete.
